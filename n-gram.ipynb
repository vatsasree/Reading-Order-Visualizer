```python
def n_gram_metric(gt,pred):
    n = len(gt)
    n_gram_score = 0
    for i in range(n):
        gt_subsequences = [gt[j:j+i+1] for j in range(len(gt)-i)]
        # print(gt_subsequences)
        pred_subsequences = [pred[j:j+i+1] for j in range(len(pred)-i)]
        # print(pred_subsequences)
        c=0
        for sub in pred_subsequences:
            if sub in gt_subsequences:
                # print(sub)
                c+=1
        print("Score for {}-gram is {}".format(i+1,c/len(gt_subsequences)))
        n_gram_score += (c/len(gt_subsequences))
    print("Final score is {}".format(n_gram_score/n))
```

```python
def broken_links_metric(gt,pred):
    gt_subsequences_2_gram = [gt[j:j+2] for j in range(len(gt)-1)]
    pred_subsequences_2_gram = [pred[j:j+2] for j in range(len(pred)-1)]
    print("GT:",gt_subsequences_2_gram)
    print("PRED:",pred_subsequences_2_gram)
    cc=0
    for seq in pred_subsequences_2_gram:
        if seq not in gt_subsequences_2_gram:
            cc+=1
            print(seq)
    print(cc)
    print("Score for broken links is {}".format(1-(cc/len(gt_subsequences_2_gram))))
    
```
